```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("SparkIceberg")
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions")
    .config("spark.sql.catalog.iceberg", "org.apache.iceberg.spark.SparkCatalog")
    .getOrCreate()
)

spark._jvm.org.apache.logging.log4j.core.config.Configurator.setLevel(
    "org.apache.iceberg.hadoop.HadoopTableOperations",
    spark._jvm.org.apache.logging.log4j.Level.ERROR,
)
```

```python
CREATE NAMESPACE IF NOT EXISTS iceberg.rh
```

```python
# Criar DataFrame
data = [
    (1, "Romerito", "Platform Engineering"),
    (2, "Theodoro", "Data Engineering"),
    (3, "Miguel", "AI Engineering"),
    (4, "Nina","Cofee Engineeing"),
    (5, "Niko","Security Engineering"),
    (6, "Leonor", "Govercance")
]

df = spark.createDataFrame(data, ["id", "nome", "area"])

display(df)
```

```python
df.write \
    .format("iceberg") \
    .mode("overwrite") \
    .saveAsTable("iceberg.rh.employees")
```

```python

# Escrever tabela Iceberg
df.writeTo("iceberg.native.employees").createOrReplace()
```
