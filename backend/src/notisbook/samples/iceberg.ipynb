```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("SparkIceberg")
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions")
    .config("spark.sql.catalog.iceberg", "org.apache.iceberg.spark.SparkCatalog")
    .getOrCreate()
)

spark._jvm.org.apache.logging.log4j.core.config.Configurator.setLevel(
    "org.apache.iceberg.hadoop.HadoopTableOperations",
    spark._jvm.org.apache.logging.log4j.Level.ERROR,
)
```

```python
CREATE NAMESPACE IF NOT EXISTS iceberg.bronze
```

```python
# Criar DataFrame
data = [
    (1, "Romerito", "Platform Engineering"),
    (2, "Theodoro", "Data Engineering"),
    (3, "Miguel", "AI Engineering"),
    (4, "Nina","Cofee Engineeing"),
    (5, "Niko","Security Engineering"),
    (6, "Leonor", "Govercance")
]

df = spark.createDataFrame(data, ["id", "nome", "area"])

display(df)
```

Gravamos o DataFrame como tabela Iceberg **externa** em `iceberg.bronze`: a propriedade `location` aponta para `s3a://notislake/bronze/` e `createOrReplace` permite reexecutar a célula.

```python
df.writeTo("iceberg.bronze.iceberg_employees") \
    .tableProperty("location", "s3a://notislake/bronze/iceberg_employees_files") \
    .createOrReplace()
```

Consultamos a tabela criada acima (célula `spark sql`).

```python
SELECT * FROM iceberg.bronze.iceberg_employees
```
