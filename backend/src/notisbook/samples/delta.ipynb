```python
from delta import configure_spark_with_delta_pip
from pyspark.sql import SparkSession

# SparkSession com Delta Lake
builder = (
    SparkSession.builder
    .appName("SparkDelta")
    .master("local[4]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
# Criar DataFrame
data = [
    (1, "Romerito", "Platform Engineering"),
    (2, "Theodoro", "Data Engineering"),
    (3, "Miguel", "AI Engineering"),
    (4, "Nina", "Coffee Engineering"),
    (5, "Niko", "Security Engineering"),
    (6, "Leonor", "Governance")
]

df = spark.createDataFrame(data, ["id", "nome", "area"])
```

```python
display(df)
```

Gravamos o DataFrame como tabela Delta **externa** na camada `bronze`: o `path` fica em `s3a://notislake/bronze/` e `overwrite` permite reexecutar a célula.

```python
(
    df.write
    .format("delta")
    .option("path", "s3a://notislake/bronze/delta_employees_files")
    .mode("overwrite")
    .saveAsTable("bronze.delta_employees")
)
```

Consultamos a tabela criada acima (célula `spark sql`).

```python
select *, current_timestamp() AS atual from bronze.delta_employees
```
