```python
from delta import configure_spark_with_delta_pip
from pyspark.sql import SparkSession

# SparkSession com Delta Lake
builder = (
    SparkSession.builder
    .appName("SparkDelta")
    .master("local[4]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
# Criar DataFrame
data = [
    (1, "Romerito", "Platform Engineering"),
    (2, "Theodoro", "Data Engineering"),
    (3, "Miguel", "AI Engineering"),
    (4, "Nina","Cofee Engineeing"),
    (5, "Niko","Security Engineering"),
    (6, "Leonor", "Govercance")
]

df = spark.createDataFrame(data, ["id", "nome", "area"])
```

```python
display(df)
```

```python
# Criar tabela Delta gerenciada
(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("delta.native_employees")
)
```

```python
df.write \
    .format("delta") \
    .option(
        "path",
        "/tmp/notisbook-samples/delta/employees"
    ) \
    .mode("overwrite") \
    .saveAsTable("delta.extern_employees")
```
