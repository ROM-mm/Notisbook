```python
from pyspark.sql import SparkSession, Row

spark = (
    SparkSession.builder
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.sql.session.timeZone", "UTC")
    .getOrCreate()
)
```

```python
# Criar DataFrame
data = [
    (1, "Romerito", "Platform Engineering"),
    (2, "Theodoro", "Data Engineering"),
    (3, "Miguel", "AI Engineering"),
    (4, "Nina","Cofee Engineeing"),
    (5, "Niko","Security Engineering"),
    (6, "Leonor", "Govercance")
]

df = spark.createDataFrame(data, ["id", "nome", "area"])

display(df)
```

```python
df.write\
    .format("hudi")\
    .option("hoodie.datasource.write.recordkey.field","id")\
    .saveAsTable("native.employees")
```

```python
df.write.format("hudi") \
    .option("hoodie.datasource.write.recordkey.field", "id") \
    .option("path", "/tmp/notisbook-samples/hudi") \
    .mode("append") \
    .saveAsTable("external.employees")
```
